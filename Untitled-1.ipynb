```python
def map_act(key_words):
    results = eidb.search(key_words)
    return results

def activity_information(activity):
    act =bd.get_activity(activity) 
    print("Descripción:", act.get('product_information'))
    information = act.as_dict()
    for clave, valor in information.items():
        print(f"{clave}: {valor}")

    return

def exchanges(activity):
    for exc in activity.exchanges():
        print(exc['amount'], exc['unit'], exc.input)
    return

def especif_exchage(word,activity):
    for exc in activity.exchanges():
        if word in exc.input['name'].lower():  # .lower() searching the flux to erase
            flux = exc.input
            print(exc['amount'], exc['unit'], exc.input['name'], exc.input['location'])

    return flux

def flux_act_df(activity):
    ca = bwa.utils.recursive_calculation_to_object(
        (activity),
        listtupla[0],
        amount=1,  # How much of the activity? (same as in our FU)
        max_level=3,  # How many levels deep in the CA do you want to go?
    )
    return pd.DataFrame(ca)

def diagramme_flux(df):
    # Crear mapeo de nodos a índices
    labels = df["label"].tolist()
    label_to_index = {label: i for i, label in enumerate(labels)}

    # Asignar source y target (ignorando filas sin parent como 'root')
    df_links = df[df["parent"].notna()].copy()
    df_links["source"] = df_links["parent"].map(label_to_index)
    df_links["target"] = df_links["label"].map(label_to_index)

    # Crear diagrama de Sankey
    fig = go.Figure(data=[go.Sankey(
        node=dict(
            pad=20,
            thickness=20,
            line=dict(color="black", width=0.5),
            label=labels,
        ),
        link=dict(
            source=df_links["source"],
            target=df_links["target"],
            value=df_links["fraction"],  # puedes usar 'amount' también
            label=df_links["name"],
        )
    )])

    fig.update_layout(title_text="Diagrama de Sankey - Producción de Acero Inoxidable", font_size=12)
    return fig.show()


def flux_replace(activity,target,option):
    for exc in activity.exchanges():
        if exc.input == target:
            print('modificando')
            exc['input'] = option.key
            exc['name'] = option['name']
            exc.save()
    return
```
